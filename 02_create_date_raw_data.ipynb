```python
from datetime import date, timedelta
from pathlib import Path
import pandas as pd

def generate_dim_date(start_date="2026-07-01", end_date="2026-08-29"):
    start = date.fromisoformat(start_date)
    end = date.fromisoformat(end_date)
    holiday_list = {
        "2026-07-01": "建党节",
        "2026-07-09": "端午节",
        "2026-08-01": "建军节",
        "2026-08-22": "七夕节",
        "2026-08-27": "中元节",
    }
    rows = []
    current = start

    while current <= end:
        iso_date = current.isoformat()
        iso_info = current.isocalendar()

        # 设置节假日，当前时间段先默认全 False
        is_holiday = 1 if iso_date in holiday_list else 0

        rows.append({
            "date_key": current.strftime("%Y%m%d"),
            "date_str": iso_date,
            "year": current.year,
            "month": current.month,
            "day": current.day,
            "week_of_year": iso_info.week,
            "day_of_week": current.isoweekday(),  # 1=Mon ... 7=Sun
            "is_weekend": 1 if current.isoweekday() >= 6 else 0,
            "is_holiday": is_holiday,
        })

        current += timedelta(days=1)

    df = pd.DataFrame(rows)

    # 统一字段顺序
    df = df[
        ["date_key", "date_str", "year", "month", "day",
         "week_of_year", "day_of_week",
         "is_weekend", "is_holiday"]
    ]

    output_dir = Path("output_data")
    output_dir.mkdir(exist_ok=True)

    output_path = output_dir / "02_dim_date.parquet"
    df.to_parquet(output_path, index=False)

    print(f"✅ 02_dim_date parquet saved to: {output_path}")
    print(f"✅ total rows: {len(df)}")
    print(df.head(10).to_string(index=False))
    return df


generate_dim_date("2026-07-01", "2026-08-29")
```
